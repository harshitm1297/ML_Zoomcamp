```python
import numpy as np
import pandas as pd

print(f"Pandas version: {pd.__version__}")
```

## Load the data

```python
data_path = "car_fuel_efficiency_2026.csv"
df = pd.read_csv(data_path)
df.shape
```

## Q2. Records count

```python
print(f"Number of records: {len(df)}")
```

**Answer: 10,000**

## Q3. Fuel types

```python
fuel_types = sorted(df["fuel_type"].dropna().unique())
print(f"Number of fuel types: {df['fuel_type'].nunique()}")
print(f"Fuel types: {fuel_types}")
```

**Answer: 3**

## Q4. Missing values

```python
missing_counts = df.isna().sum()
print(missing_counts[missing_counts > 0])
print(f"Number of columns with missing values: {df.isna().any().sum()}")
```

**Answer: 2**

## Q5. Maximum fuel efficiency for cars from Asia

```python
asia_max_mpg = df.loc[df["origin"] == "Asia", "fuel_efficiency_mpg"].max()
print(f"Maximum fuel efficiency for Asia: {asia_max_mpg} MPG")
```

**Answer: 41.2**

## Q6. Median horsepower before and after mode imputation

```python
median_before = df["horsepower"].median()
horsepower_mode = df["horsepower"].mode().iloc[0]
horsepower_filled = df["horsepower"].fillna(horsepower_mode)
median_after = horsepower_filled.median()

print(f"Median before filling: {median_before}")
print(f"Most frequent value: {horsepower_mode}")
print(f"Median after filling: {median_after}")
print("Result: Yes, it decreased" if median_after < median_before else "Result: It did not decrease")
```

**Answer: Yes, it decreased**

## Q7. Linear algebra

```python
X = (
    df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]]
      .head(7)
      .to_numpy()
)
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print("X:")
print(X)
print("\nXTX:")
print(XTX)
print(f"\nw: {w}")
print(f"Sum of w: {w.sum()}")
```

**Answer: 0.369** (unrounded value: `0.36919696904925486`)
